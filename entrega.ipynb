# Entrega - Trial-Based Online Policy Improvement

Este notebook presenta el estudio experimental del agente de Connect-4. El objetivo es mostrar que el desempeno no es un valor unico y atomico, sino que depende de variables como la cantidad de trials, el tipo de rollout, el oponente y el orden de juego.

**Agente final:** `Trial-Based Online Policy Improvement con rollouts heuristicos`.  
**Version base:** la misma tecnica de mejora online, pero con rollouts aleatorios.  
**Baselines y oponentes:** `Random`, la version con rollouts aleatorios y self-play.


## Idea principal del agente

El agente final usa **Trial-Based Online Policy Improvement**. En cada turno revisa las acciones legales, simula partidas completas desde cada accion candidata y estima el valor promedio de esa accion. Despues juega la columna con mejor estimacion.

La diferencia principal frente a la version base esta en la politica usada dentro de cada simulacion:

- **Rollouts aleatorios:** durante los trials se eligen movimientos legales al azar. Es una version rapida y simple, pero sus estimaciones tienen mas ruido.
- **Rollouts heuristicos:** durante los trials se usa conocimiento de Connect-4: ganar si es posible, bloquear victorias inmediatas, priorizar columnas centrales, construir amenazas y evitar entregar jugadas ganadoras.

La hipotesis del estudio es que los rollouts heuristicos producen mejores decisiones con menos trials, aunque cada trial sea mas costoso computacionalmente.

## Conclusiones

1. **La heuristica mejora la calidad de los rollouts.** Trial-Based Online Policy Improvement con rollouts heuristicos alcanza 100% de victorias contra `Random` desde 1 trial, mientras la version con rollouts aleatorios necesita mas trials para estabilizarse.
2. **La mejora tambien aparece en la comparacion directa.** La version heuristica vence con frecuencia a la version con rollouts aleatorios, lo que indica que la heuristica aporta mas que solo explotar debilidades del jugador aleatorio.
3. **El costo computacional es la debilidad principal.** El agente heuristico tarda mas por jugada porque cada paso simulado requiere evaluacion heuristica.
4. **El orden de juego importa.** Por eso las graficas separan primer y segundo jugador o quien inicia, como pide el analisis por variables experimentales.
5. **Self-play aporta estabilidad al analisis.** Permite observar si el agente se comporta razonablemente contra una copia de si mismo y si hay ventaja del primer jugador.


## Debilidades y mejoras futuras

- **Debilidad:** Trial-Based Online Policy Improvement con rollouts heuristicos puede ser lento con muchos trials.  
  **Mejora:** usar asignacion adaptativa de trials, deteniendo temprano acciones claramente inferiores.

- **Debilidad:** algunas configuraciones de comparacion directa tienen alta varianza por tener pocas partidas.  
  **Mejora:** aumentar partidas solo en los puntos criticos, por ejemplo trials 25 y 50.

- **Debilidad:** el rollout heuristico puede perder diversidad si siempre favorece las mismas acciones.  
  **Mejora:** mantener muestreo estocastico entre las mejores acciones y ajustar la temperatura de seleccion.

- **Debilidad:** el agente puede gastar tiempo en estados ya tacticamente claros.  
  **Mejora:** detectar victorias, bloqueos y amenazas forzadas antes de lanzar todos los trials.
