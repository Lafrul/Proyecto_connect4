# Luis Value Policy Agent

This notebook evaluates the three versions of the agent against a random player and against each other.

```python
from groups.Luis.gui import RandomAgent, play_match
from groups.Luis import policy
import matplotlib.pyplot as plt

agents = {
    "Random": RandomAgent,
    "Luis Tactical": policy.LuisTactical,
    "Luis Value Random": policy.LuisValueRandom,
    "Luis Value Self Play": policy.LuisValueSelfPlay,
}

games = 100
```

```python
results = []

for name in ["Luis Tactical", "Luis Value Random", "Luis Value Self Play"]:
    red_wins, yellow_wins, draws = play_match(agents[name], agents["Random"], games)
    results.append({"agent": name, "color": "red", "wins": red_wins, "losses": yellow_wins, "draws": draws})

    red_wins, yellow_wins, draws = play_match(agents["Random"], agents[name], games)
    results.append({"agent": name, "color": "yellow", "wins": yellow_wins, "losses": red_wins, "draws": draws})

results
```

```python
labels = [item["agent"] + " " + item["color"] for item in results]
wins = [item["wins"] for item in results]
losses = [item["losses"] for item in results]
draws = [item["draws"] for item in results]

plt.figure(figsize=(12, 5))
plt.bar(labels, wins, label="wins")
plt.bar(labels, losses, bottom=wins, label="losses")
plt.bar(labels, draws, bottom=[wins[i] + losses[i] for i in range(len(wins))], label="draws")
plt.xticks(rotation=30, ha="right")
plt.ylabel("games")
plt.title("Agent performance against Random")
plt.legend()
plt.show()
```

```python
matrix = []
names = ["Random", "Luis Tactical", "Luis Value Random", "Luis Value Self Play"]

for red in names:
    for yellow in names:
        if red == yellow:
            continue
        red_wins, yellow_wins, draws = play_match(agents[red], agents[yellow], 40)
        matrix.append((red, yellow, red_wins, yellow_wins, draws))

matrix
```
